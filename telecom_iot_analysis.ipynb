```python
# Import pandas library
import pandas as pd

# Empty list to store processed chunk summaries
customer_revenue = []

# Read revenue dataset in chunks
for chunk in pd.read_csv(
    "data/archive (4)/rev1.csv",
    chunksize=50000
):

    # Calculate revenue statistics for each customer
    summary = chunk.groupby("msisdn").agg({
        "revenue_usd": ["mean", "max", "min", "sum"]
    })

    # Rename columns
    summary.columns = [
        "avg_revenue",
        "max_revenue",
        "min_revenue",
        "total_revenue"
    ]

    # Convert index to column
    summary.reset_index(inplace=True)

    # Store processed chunk
    customer_revenue.append(summary)

# Combine all chunk summaries into one dataframe
revenue_summary = pd.concat(
    customer_revenue,
    ignore_index=True
)

print("Revenue Summary Shape:")
print(revenue_summary.shape)
# Read the large revenue dataset in chunks
# Generate revenue statistics for each customer
# Combine all processed chunks into a single dataframe
# This approach reduces memory usage and supports large-scale telecom data processing
```

```python
# Combine duplicate customers created across chunks
# After processing multiple chunks, the same customer may appear
# in more than one chunk. Therefore, a final aggregation step is
# performed to create one record per customer.
final_revenue = revenue_summary.groupby("msisdn").agg({
    "avg_revenue": "mean",
    "max_revenue": "max",
    "min_revenue": "min",
    "total_revenue": "sum"
}).reset_index()

print("Final Revenue Shape:")
print(final_revenue.shape)

print("\nUnique Customers:")
print(final_revenue["msisdn"].nunique())
```

```python
# Create a set of customer IDs from revenue table
# A set provides very fast lookup performance
# Extract all unique customer identifiers from the
# revenue summary table. These customers will be used
# to filter the large CRM dataset.
revenue_customers = set(
    final_revenue["msisdn"]
)

print("Revenue Customers:")
print(len(revenue_customers))
```

```python
# Display first few customer IDs

list(revenue_customers)[:5]
```

```python
# Import pandas
import pandas as pd

# Store filtered CRM records
crm_filtered = []

# Read CRM dataset in chunks
for chunk in pd.read_csv(
    "data/archive (4)/crm1.csv",
    chunksize=50000
):

    # Keep only customers that exist
    # in the revenue table
    filtered_chunk = chunk[
        chunk["msisdn"].isin(revenue_customers)
    ]

    # Store matching customers
    crm_filtered.append(filtered_chunk)

# Combine all matching CRM chunks
crm_final = pd.concat(
    crm_filtered,
    ignore_index=True
)

print("Filtered CRM Shape:")
print(crm_final.shape)
# Read CRM dataset in chunks to avoid loading
# 13.6 million records into memory at once

# Filter only customers that appear in
# the revenue summary table

# Combine matching CRM records into
# a final customer profile dataset
```

```python
# Read Device dataset in chunks
# Keep only customers present in revenue summary
# Store matching device records

# Combine all filtered chunks into a single dataset
device_filtered = []

for chunk in pd.read_csv(
    "data/archive (4)/device1.csv",
    chunksize=50000
):

    filtered_chunk = chunk[
        chunk["msisdn"].isin(revenue_customers)
    ]

    device_filtered.append(filtered_chunk)

# Combine filtered device records
device_final = pd.concat(
    device_filtered,
    ignore_index=True
)

print("Filtered Device Shape:")
print(device_final.shape)
```

```python
# Merge revenue summary with CRM data
# Merge customer revenue features with CRM profile data
# using msisdn as the common customer identifier
final_df = pd.merge(
    final_revenue,
    crm_final,
    on="msisdn",
    how="left"
)

print("After CRM Merge:")
print(final_df.shape)
```

```python
# Merge device information
# Merge device information to create a unified
# telecom customer dataset containing revenue,
# customer profile and device attributes
final_df = pd.merge(
    final_df,
    device_final,
    on="msisdn",
    how="left"
)

print("Final Dataset Shape:")
print(final_df.shape)
```

```python
# Keep only one record per customer

crm_final_clean = crm_final.drop_duplicates(
    subset="msisdn"
)

print("CRM Clean Shape:")
print(crm_final_clean.shape)
```

```python
# Merge revenue and cleaned CRM data

final_df = pd.merge(
    final_revenue,
    crm_final_clean,
    on="msisdn",
    how="left"
)

print("After Clean CRM Merge:")
print(final_df.shape)
```

```python
# Remove duplicate device records
# Keep one device profile per customer

device_final_clean = device_final.drop_duplicates(
    subset="msisdn"
)

print("Device Clean Shape:")
print(device_final_clean.shape)
```

```python
# Merge device information with final dataset

final_df = pd.merge(
    final_df,
    device_final_clean,
    on="msisdn",
    how="left"
)

print("Final Dataset Shape:")
print(final_df.shape)

print("\nUnique Customers:")
print(final_df["msisdn"].nunique())
```

```python
# Check missing values in final dataset

print(final_df.isnull().sum())
```

```python
# Create customer age from year of birth

final_df["age"] = 2025 - final_df["year_of_birth"]

final_df[[
    "year_of_birth",
    "age"
]].head()
```

```python
# Fill missing categorical values

final_df["gender"] = final_df["gender"].fillna("Unknown")

final_df["system_status"] = final_df["system_status"].fillna("Unknown")

final_df["mobile_type"] = final_df["mobile_type"].fillna("Unknown")

final_df["value_segment"] = final_df["value_segment"].fillna("Unknown")

final_df["brand_name"] = final_df["brand_name"].fillna("Unknown")

final_df["model_name"] = final_df["model_name"].fillna("Unknown")

final_df["os_name"] = final_df["os_name"].fillna("Unknown")

final_df["os_vendor"] = final_df["os_vendor"].fillna("Unknown")
```

```python
# Create revenue drop percentage
# Identify customers whose revenue has significantly
# dropped compared to their historical performance.
final_df["revenue_drop"] = (
    final_df["max_revenue"]
    - final_df["avg_revenue"]
)

# Calculate threshold using top 10% revenue drops

threshold = final_df["revenue_drop"].quantile(0.90)

print("Revenue Drop Threshold:")
print(threshold)
```

```python
# Create risk labels
# Customers with large revenue drops are classified
# as High Risk customers.
final_df["risk_level"] = final_df["revenue_drop"].apply(
    lambda x:
    "High Risk"
    if x > threshold
    else "Normal"
)

print(final_df["risk_level"].value_counts())
```

```python
# Create outreach customer segment
# High Risk customers are marked as Outreach Customers
# so they can be targeted with retention campaigns.
final_df["customer_segment"] = final_df["risk_level"].apply(
    lambda x:
    "Outreach Customer"
    if x == "High Risk"
    else "Regular Customer"
)

print(final_df["customer_segment"].value_counts())
```

```python
# Identify enterprise customers
# Customers in the top 25% of total revenue
# are classified as Enterprise Customers.
final_df["enterprise_customer"] = (
    (final_df["total_revenue"] > final_df["total_revenue"].quantile(0.75))
)

print(final_df["enterprise_customer"].value_counts())
```

```python
# Function to recommend offers
# Generate personalized retention offers
# based on customer risk and value segment.
def recommend_offer(row):

    if row["risk_level"] == "High Risk":

        if row["value_segment"] == "Tier_1":
            return "Premium Loyalty Offer"

        elif row["value_segment"] == "Tier_2":
            return "Revenue Retention Offer"

        else:
            return "Discount Retention Offer"

    else:
        return "No Offer Needed"


# Apply recommendation logic

final_df["recommended_offer"] = final_df.apply(
    recommend_offer,
    axis=1
)

print(final_df["recommended_offer"].value_counts())
```

```python
# Generate personalized email text
# Generate automated customer emails
# based on risk level and recommended offer.
def create_email(row):

    if row["risk_level"] == "High Risk":

        return (
            f"Dear Customer, "
            f"We noticed a decline in your recent revenue activity. "
            f"As a valued customer, we are offering you a "
            f"{row['recommended_offer']}. "
            f"Thank you for choosing our telecom services."
        )

    else:

        return (
            "Thank you for being a valued customer."
        )


# Create email column

final_df["email_text"] = final_df.apply(
    create_email,
    axis=1
)

print(final_df[[
    "risk_level",
    "recommended_offer",
    "email_text"
]].head())
```

```python
# Select final output columns
# Create final customer scorecard containing
# revenue metrics, risk labels, offers and emails.
final_scorecard = final_df[[
    "msisdn",
    "avg_revenue",
    "total_revenue",
    "risk_level",
    "customer_segment",
    "enterprise_customer",
    "recommended_offer",
    "email_text"
]]

print(final_scorecard.shape)

final_scorecard.head()
```

```python
# Save final scorecard
# Export final scorecard for dashboard integration.
final_scorecard.to_csv(
    "outputs/final_customer_scorecard.csv",
    index=False
)

print("Final Scorecard Saved Successfully")
```
